- conditional expression - `value_if_true if condition else value_if_false`


```python
x=2

result = 'x is greater than 5' if x > 5 else 'x is 5 or smaller'
print(result)
```

### For Loop

- syntax - `for num in number:`

```python
numbers = [1,2,3,4,5]

for num in numbers:
    print(f'Number is {num}')
```

```python
for i in range(5):
    print(f'i is: {i}')
```

```python
for i in range(1,20,2):
    print(f'i is: {i}')
```

```python
for i in range(10):
    if i % 2 ==0:
        print(f'{i} is even')
```

```python
rain
```

```python
total = 0
for r in rain:
    total = total + r

print(f'Total rainfall: {round(total, 1)} mm')
```
